## 3. Give the model instructions (a system prompt)

A system prompt tells the model how to behave. Keep prompts you will report in files (see the `prompts/` folder) so they are saved with your project.

```python
from pathlib import Path

system = Path("../prompts/system_coding.md").read_text()
user_template = Path("../prompts/code_excerpt.md").read_text()
excerpt = "Honestly the AI saves me time on boilerplate, but I no longer trust I understand my own code."

reply = llm.chat(user_template.format(speaker="P04", text=excerpt), system=system)
print(reply.text)
```

## 4. Get a structured answer

A JSON schema forces the answer into a fixed shape, which is handy for tables. The result below is shown as a table using pandas.

```python
import json
import pandas as pd

schema = json.loads(Path("../prompts/schema.json").read_text())
reply = llm.chat(user_template.format(speaker="P04", text=excerpt), system=system, json_schema=schema)
pd.DataFrame([reply.json()])
```

## 5. Privacy check

Before anything is sent, the lab's tools check your text for names, emails and other identifiers and refuse to send it if they find some.
To check a file yourself, open a terminal and run `lab scan <file>`.

```python
from labllm import IdentifierError

fake_email = "jane.doe" + "@" + "example.edu"   # built here so it isn't flagged when you save a version
try:
    llm.chat(f"Summarize: Jane ({fake_email}) said the lab was stressful.")
except IdentifierError as e:
    print("Blocked as expected:", e)
```
